```python
import networkx as nx
```

```python
dataset = 'Brexit'
file_path = f'../../input/datasets/static/{dataset}/edgelist_dith'
output_path = f'./dith/datasets/{dataset}/'
```

```python
# read in nx.graph from the file, where the first line is n, m, A, R, and the following m lines(edges) are u, v, d, w
# n is #Nodes, m is #Edges, A is attractor node, R is repulser node, u, v are the nodes of the edge, d is the distance, w is the edge weight
with open(file_path, 'r') as f:
    first_line = f.readline().strip().split()
    n = int(first_line[0])  # number of nodes
    m = int(first_line[1])  # number of edges
    A = int(first_line[2])  # attractor node
    R = int(first_line[3])  # repulser node
    
    # Create a directed graph
    G = nx.Graph()
    
    # Read the edges and add them to the graph
    for _ in range(m):
        line = f.readline().strip().split()
        u = int(line[0])  # source node
        v = int(line[1])  # target node
        d = float(line[2])  # distance
        w = float(line[3])  # edge weight
        G.add_edge(u, v, distance=d, weight=w)
```

```python
d_A = dict(nx.shortest_path_length(G, A, weight='distance'))
max_d_A = max(d_A.values())
d_A_rev = {v: max_d_A - d_A[v] for v in d_A}

d_R = dict(nx.shortest_path_length(G, R, weight='distance'))
max_d_R = max(d_R.values())
d_R_rev = {v: max_d_R - d_R[v] for v in d_R}
```

```python
# check the connectivity of the graph
if not nx.is_connected(G):
    print("The graph is not connected")
else:
    print("The graph is connected")
```

```python
# output 5 files to the output_path:
# 1. graph.tsv: m lines, each line is u v w
# 2. d_A.tsv: n lines, each line is v d(key and value in d_A)
# 3. d_A_rev.tsv: n lines, each line is v d(key and value in d_A_rev)
# 4. d_R.tsv: n lines, each line is v d(key and value in d_R)
# 5. d_R_rev.tsv: n lines, each line is v d(key and value in d_R_rev)
import os

# Ensure the output directory exists
os.makedirs(output_path, exist_ok=True)

# 1. Write graph.tsv: m lines, each line is u v w
with open(os.path.join(output_path, 'graph.tsv'), 'w') as f_graph:
    for u, v, data in G.edges(data=True):
        f_graph.write(f"{u}\t{v}\t{data['weight']}\n")

# 2. Write d_A.tsv: n lines, each line is v d (from d_A)
with open(os.path.join(output_path, 'd_A.tsv'), 'w') as f_d_A:
    for v in range(n):
        d = d_A.get(v, 'inf')
        f_d_A.write(f"{v}\t{d}\n")

# 3. Write d_A_rev.tsv: n lines, each line is v d (from d_A_rev)
with open(os.path.join(output_path, 'd_A_rev.tsv'), 'w') as f_d_A_rev:
    for v in range(n):
        d = d_A_rev.get(v, 'inf')
        f_d_A_rev.write(f"{v}\t{d}\n")

# 4. Write d_R.tsv: n lines, each line is v d (from d_R)
with open(os.path.join(output_path, 'd_R.tsv'), 'w') as f_d_R:
    for v in range(n):
        d = d_R.get(v, 'inf')
        f_d_R.write(f"{v}\t{d}\n")

# 5. Write d_R_rev.tsv: n lines, each line is v d (from d_R_rev)
with open(os.path.join(output_path, 'd_R_rev.tsv'), 'w') as f_d_R_rev:
    for v in range(n):
        d = d_R_rev.get(v, 'inf')
        f_d_R_rev.write(f"{v}\t{d}\n")
```
