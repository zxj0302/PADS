```python
%load_ext autoreload
%autoreload 2
from RePBubLik import RePBubLik
import matplotlib.pyplot as plt
import networkx as nx
```

```python
graph_file = '../../../output/results-theta=0.5/Referendum_/graph.gml'
G = nx.read_gml(graph_file)
G.remove_edges_from(nx.selfloop_edges(G))
# get the largest connected component
largest_cc = max(nx.connected_components(G), key=len)
G = G.subgraph(largest_cc).copy()
# convert node labels to integers
mapping = {node: int(node) for node in list(G.nodes())}
G = nx.relabel_nodes(G, mapping)

new_edges = RePBubLik(G=G, t=10, b=2, topk=10, maxedges=100, unweighted=True, iter=1)
print(new_edges)
```
