```python
from modules.ps import ge
import networkx as nx
```

```python
G = nx.read_gml('../../output/results-theta=0.5/Referendum_/graph.gml')
# get the largest connected component
largest_cc = max(nx.connected_components(G), key=len)
G = G.subgraph(largest_cc).copy()
src = {i: G.nodes[i]['polarity'] for i in G.nodes() if G.nodes[i]['polarity'] > 0}
trg = {i: G.nodes[i]['polarity'] for i in G.nodes() if G.nodes[i]['polarity'] < 0}
res = ge(src, trg, G)
print(res)
```

```python
# Add 1000 edges to G
import random
for i in range(1000):
    G.add_edge(random.choice(list(G.nodes())), random.choice(list(G.nodes())))
res = ge(src, trg, G)
print(res)
```
